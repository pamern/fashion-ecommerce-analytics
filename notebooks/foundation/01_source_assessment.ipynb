# 01. Data Inventory

This notebook provides a high-level inventory of the source datasets, including file names, row counts, column counts, and file sizes.

```python
from pathlib import Path

import pandas as pd


# Xác định project root dù notebook được mở từ root hay notebooks/
project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

raw_dir = project_root / "data" / "raw"

print(f"Raw data directory: {raw_dir}")
```

```python
inventory = []

for file_path in sorted(raw_dir.glob("*.csv")):
    df = pd.read_csv(file_path, low_memory=False)

    inventory.append({
        "file_name": file_path.name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "size_mb": round(file_path.stat().st_size / (1024 ** 2), 2),
    })

data_inventory = pd.DataFrame(inventory)

data_inventory
```
